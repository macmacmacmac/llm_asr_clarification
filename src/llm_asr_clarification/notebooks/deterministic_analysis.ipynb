```python
for m in dataset1:
        eq_45 = (
            torch.equal(dataset4[m]["logps"], dataset5[m]["logps"]),
            torch.equal(dataset4[m]["segment_deltas"], dataset5[m]["segment_deltas"]),
        )
        max_logp_diff = (dataset1[m]["logps"] - dataset4[m]["logps"]).abs().max().item()
        max_delta_diff = (dataset1[m]["segment_deltas"] - dataset4[m]["segment_deltas"]).abs().max().item()
        print(f"{m}: det4==det5={eq_45} | max |logp_1 - logp_4|={max_logp_diff:.3e} | max |delta_1 - delta_4|={max_delta_diff:.3e}")
```
